```python
df = df.drop_duplicates()
```

```python
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
```

```python
df.info()
```

```python
total_sales = df['Sales_Amount'].sum()
print("Total Sales:", total_sales)
```

```python
average_sales = df['Sales_Amount'].mean()
print("Average Sales:", average_sales)
```

```python
maximum_sales = df['Sales_Amount'].max()
print("Maximum Sale:", maximum_sales)
```

```python
minimum_sales = df['Sales_Amount'].min()
print("Minimum Sale:", minimum_sales)
```

```python
product_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)

print(product_sales)
```

```python
product_sales.head(10)
```

```python
product_sales.tail(10)
```

```python
plt.figure(figsize=(12,6))

product_sales.head(10).plot(kind='bar')

plt.title('Top 10 Products by Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Month'] = df['Sale_Date'].dt.to_period('M')
```

```python
monthly_sales = df.groupby('Month')['Sales_Amount'].sum()
```

```python
print(monthly_sales)
```

```python
plt.figure(figsize=(14,6))

monthly_sales.plot()

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Month_Name'] = df['Sale_Date'].dt.month_name()
```

```python
monthly_average = df.groupby('Month_Name')['Sales_Amount'].mean()
```

```python
df['Month_Number'] = df['Sale_Date'].dt.month

monthly_sales = df.groupby('Month_Number')['Sales_Amount'].sum()

monthly_sales
```

```python
plt.figure(figsize=(10,6))

monthly_sales.plot(kind='bar')

plt.title('Monthly Sales Performance')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(
    range(12),
    ['Jan','Feb','Mar','Apr','May','Jun',
     'Jul','Aug','Sep','Oct','Nov','Dec'],
    rotation=45
)

plt.tight_layout()
plt.show()
```

```python
quantity_by_product = df.groupby('Product_ID')['Quantity_Sold'].sum().sort_values(ascending=False)

print(quantity_by_product)
```

```python
plt.figure(figsize=(12,6))

quantity_by_product.head(10).plot(kind='bar')

plt.title('Top 10 Products by Quantity Sold')
plt.xlabel('Product')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)

print(category_sales)
```

```python
plt.figure(figsize=(10,6))

category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Product_Category')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
plt.figure(figsize=(10,6))

region_sales.plot(kind='bar')

plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.select_dtypes(include='number').corr(),
            annot=True,
            cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
